```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scripts.data_cleaning import load_data, clean_data, save_cleaned


df = load_data("data/benin.csv")


display(df.describe())
display(df.isna().sum())


df_clean = clean_data(df)


save_cleaned(df_clean, "data/benin_clean.csv")


plt.figure(figsize=(12,5))
plt.plot(df_clean['Timestamp'], df_clean['GHI'], label='GHI')
plt.plot(df_clean['Timestamp'], df_clean['DNI'], label='DNI')
plt.plot(df_clean['Timestamp'], df_clean['DHI'], label='DHI')
plt.xlabel('Timestamp')
plt.ylabel('Solar Radiation (W/m²)')
plt.title('Solar Radiation over Time')
plt.legend()
plt.show()


plt.figure(figsize=(10,8))
sns.heatmap(df_clean[['GHI','DNI','DHI','TModA','TModB']].corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()


cleaning_avg = df_clean.groupby('Cleaning')[['ModA','ModB']].mean()
cleaning_avg.plot(kind='bar', figsize=(6,4), title='Pre/Post Cleaning Module Performance')
plt.show()
```
